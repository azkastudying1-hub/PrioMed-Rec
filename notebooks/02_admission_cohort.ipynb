# 02 — Admission Cohort

**PrioMed-Rec | Sub-Task 3**

Defines and builds the admission cohort: encounters classified as IMP, EMER, or HH.
Output: `data/interim/admission_cohort.csv`

```python
import sys
sys.path.insert(0, '..')
import pandas as pd
from src.data.admission_cohort import build_admission_cohort, save_admission_cohort, ADMISSION_CLASSES
print('Admission classes:', ADMISSION_CLASSES)
```

## 1. Build and inspect cohort

```python
cohort = build_admission_cohort('../data/raw/synthea')
print(f'Cohort: {len(cohort):,} rows | {cohort["patient_id"].nunique():,} unique patients')
cohort.head()
```

## 2. Class distribution

```python
dist = cohort['encounter_class'].value_counts()
print(dist.to_string())
print(f'\nTotal excluded (AMB + VR): {488422 - len(cohort):,}')
```

## 3. Date range and stay duration

```python
cohort['stay_hours'] = (cohort['encounter_end'] - cohort['encounter_start']).dt.total_seconds() / 3600
print('Stay duration (hours):')
print(cohort['stay_hours'].describe().round(2).to_string())
print(f'\nEncounter date range: {cohort["encounter_start"].min()} to {cohort["encounter_start"].max()}')
```

## 4. Age distribution in cohort

```python
print('Age distribution:')
print(cohort['age'].describe().round(1).to_string())
print(f'\nAge 65+ rate: {cohort["age_65_plus"].mean()*100:.1f}%')
```

## 5. Save cohort

```python
save_admission_cohort(cohort, '../data/interim/admission_cohort.csv')
print('Done.')
```
